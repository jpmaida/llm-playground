## Como criar um ambiente de desenvolvimento Python?

Um ambiente virtual é uma pasta isolada que contém uma cópia do interpretador Python e as bibliotecas necessárias para um projeto específico. Ele é muito útil porque evita conflitos entre dependências de diferentes projetos.

### Por que usar ambientes virtuais?

- Cada projeto pode ter suas próprias versões de bibliotecas.
- Você evita que uma instalação global afete outros projetos.
- Facilita a reprodução do ambiente em outra máquina.

### Criando um ambiente virtual

No terminal, você pode criar um ambiente virtual com o comando:

```bash
python -m venv .venv
```

Esse comando cria uma pasta chamada `.venv` com o ambiente isolado.

> 💡 **Dica**
>
> É interessante você não fazer _commit_ desta pasta no seu repositório git pois além de ser pesada representa configurações do seu ambiente local. Você não vai querer isso exposto na internet, né!?
>
> Para que o git ignore esse diretório, crie o arquivo oculto **.gitignore** na raiz do seu repositório git local e adicione o caminho para este diretório.

### Ativando o ambiente virtual

No Linux ou macOS 💻:

```bash
source .venv/bin/activate
```

No Windows 🪟:

```bash
.venv\Scripts\activate
```

Depois de ativado, o prompt do terminal normalmente muda para mostrar o nome do ambiente, por exemplo:

```bash
(.venv) usuario@maquina:~$
```

### Exemplo prático

Você pode verificar se o ambiente foi ativado executando:

```bash
python --version
```

Ou listar os pacotes instalados:

```bash
pip list
```

> Dica: sempre ative o ambiente virtual antes de instalar bibliotecas com `pip`.


## Como executar um programa Python?

Uma forma simples de criar um script Python dentro do notebook é usar a magia 🪄 `%%writefile`. Isso faz com que o conteúdo da célula seja salvo em um arquivo com o nome que você indicar.

### Criando um arquivo Python

Execute a célula abaixo para criar um arquivo chamado `ola_mundo.py` com um programa simples:

````python
%%writefile ola_mundo.py
print("Olá, mundo!")
print("Este é um programa simples em Python.")
````

### Como executar o programa

Depois de salvar o arquivo, você pode executá-lo de duas formas:

1. No terminal, dentro da pasta do projeto:

```bash
python ola_mundo.py
```

2. Dentro do próprio notebook, usando o comando:

```python
!python ola_mundo.py
```

### O que esse programa faz?

Ele imprime mensagens na tela. Esse tipo de exemplo é útil para testar se o Python está instalado corretamente e se o ambiente virtual está funcionando.

### Boas práticas

- Use nomes descritivos para os arquivos, como `meu_script.py`.
- Organize seu código em pastas quando o projeto crescer.
- Sempre teste seus scripts após criar ou modificar o arquivo.

````python
!python ola_mundo.py
````

## Como instalar bibliotecas em Python?

Quando você trabalha com Python, muitas vezes precisa instalar bibliotecas externas para adicionar funcionalidades ao seu projeto. O comando mais recomendado para isso é:

```bash
python -m pip install nome_da_biblioteca
```

### Se o pip não estiver instalado

Em algumas máquinas, o `pip` pode não estar disponível por padrão. Nesses casos, você pode instalá-lo com:

**No Linux/macOS:**

* Fedora/RHEL:

```bash
sudo dnf install python3-pip
```

* Ubuntu / Debian / Linux Mint:

```bash
sudo apt update && sudo apt install python3-pip
```

* MacOS
Geralmente já vem instalado no sistema operacional.

```bash
python3 -m ensurepip --upgrade
```

**No Windows:**

```bash
python -m ensurepip --default-pip
```

Depois disso, você pode verificar se o `pip` foi instalado corretamente com:

```bash
python -m pip --version
```

### Exemplo

Por exemplo, para instalar a biblioteca `requests`, você pode executar:

```bash
python -m pip install requests
```

ou

```bash
pip install requests
```

Depois da instalação, você pode verificar se a biblioteca está disponível com:

```bash
python -m pip show requests
```

### Como isso se comunica com o ambiente virtual?

Quando você ativa um ambiente virtual, o comando `python` passa a apontar para o interpretador desse ambiente. Isso significa que qualquer biblioteca instalada com `pip` será colocada dentro da pasta do ambiente virtual, e não no Python global do sistema.

Em outras palavras:

- o ambiente virtual isolado mantém suas próprias bibliotecas;
- cada projeto pode ter versões diferentes das mesmas bibliotecas;
- isso evita conflitos entre projetos.

Você também pode listar as bibliotecas instaladas no ambiente ativo com:

```bash
pip list
```

Se quiser reproduzir o mesmo ambiente em outro computador, pode salvar as dependências em um arquivo chamado `requirements.txt`:

```bash
python -m pip freeze > requirements.txt
```

ou 

```bash
pip freeze > requirements.txt
```

E, depois, instalar tudo novamente com:

```bash
python -m pip install -r requirements.txt
```
