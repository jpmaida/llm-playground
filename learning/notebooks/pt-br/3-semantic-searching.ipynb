**Como resolve este problema ?**

## Busca léxica vs. Busca semântica

A busca lexical é um método tradicional de recuperação de informações que localiza documentos ou dados comparando palavras, frases ou tokens exatos de uma consulta do usuário com um índice. Em vez de interpretar o significado conceitual ou o contexto de uma frase, ela se baseia na comparação literal de texto e na pontuação com base na frequência dos termos.

Busca semântica (*Semantic Search*) é uma técnica de recuperação de informação que procura documentos com base no **significado** da consulta, e não apenas nas palavras utilizadas.

Ao invés de perguntar:

> "Esse documento contém exatamente esta palavra?"

A busca semântica pergunta:

> "Esse documento possui um significado semelhante ao da consulta?"

Isso permite encontrar documentos relevantes mesmo quando utilizam palavras diferentes, sinônimos ou formas distintas de expressar uma mesma ideia.

Ok ... já entendemos que necessitamos de uma busca que entenda o contexto da nossa pergunta original, mas como uma máquina vai entender o contexto de uma pergunta se essa é característica de um ser humano? A resposta é: **vetores**.

## O que são embeddings?

Computadores não entendem texto da mesma forma que seres humanos.

Antes que seja possível comparar frases ou documentos, eles precisam ser convertidos para uma representação numérica.

Essa representação é chamada de **Embedding**.

Um embedding é um vetor de números reais capaz de capturar o significado semântico de um texto.

Por exemplo:

**"Luke Skywalker"**

↓

**[-0.214, 0.532, -0.781, ..., 0.194]**

Na prática, modelos modernos produzem vetores com centenas ou milhares de dimensões.

Esses vetores são construídos de forma que textos semanticamente semelhantes fiquem próximos uns dos outros no espaço vetorial.

### Embeddings representam significado

Considere os seguintes textos:

- Luke Skywalker
- Yoda
- Darth Vader
- Pizza

Após serem convertidos em embeddings, esperamos que:

- Luke e Yoda estejam relativamente próximos.
- Luke e Darth Vader também apresentem alguma proximidade.
- Pizza esteja distante dos demais.

Isso acontece porque o modelo aprendeu relações semânticas durante seu treinamento, mesmo sem receber regras explícitas sobre Star Wars.

## Busca Semântica × Embeddings

É comum confundir esses dois conceitos, mas eles representam partes diferentes da solução.

### Embeddings

São uma representação vetorial de um texto.

Seu objetivo é transformar linguagem natural em números preservando o significado semântico.

### Busca Semântica

É o processo de utilizar esses vetores para localizar documentos semanticamente semelhantes.

Em outras palavras:

- Os **embeddings** são os dados.
- A **busca semântica** é o algoritmo que utiliza esses dados.

Um não substitui o outro.

Na maioria das aplicações modernas, a busca semântica depende da existência de embeddings previamente calculados.

## Como meu modelo de embeddings consegue saber dois assuntos são relacionados ?

Como dito anteriormente, em uma busca semântica a palavra "Luke" e "Yoda" são semanticamente relacionadas, mesmo que não compartilhem palavras em comum. Isso é possível porque o modelo de embeddings foi treinado em grandes quantidades de texto, aprendendo padrões e relações entre palavras e conceitos. Entretanto, como essa similaridade é calculada ?

### Similaridade por cosseno

Diferentemente de outras métricas, ela **não compara a distância entre os vetores**, mas sim **o ângulo formado entre eles**.

Quanto menor esse ângulo, maior a similaridade semântica entre os textos.

![Similaridade por cosseno](https://miro.medium.com/v2/resize:fit:2000/1*D7DdTbHwtnMXtyy1-mfxzA.png)

Se os vetores apontam praticamente para a mesma direção, o cosseno do ângulo será próximo de **1**, indicando alta similaridade.

Se forem perpendiculares, o valor será próximo de **0**, indicando pouca relação.

Se apontarem para direções opostas, o valor será **-1**, indicando significados completamente diferentes.

Matematicamente:

$$
\text{CosineSimilarity}(A, B) = \cos(\theta) = \frac{A \cdot B}{\|A\| \|B\|}
$$

onde:

- **A · B** representa o produto escalar;
- **||A||** representa o comprimento (norma) do vetor A;
- **||B||** representa o comprimento (norma) do vetor B.

Destrinchando um pouco mais a fórmula temos:

$$
\frac{A \cdot B}{\|A\| \|B\|} = \frac{\sum_{i=1}^{n} A_i B_i}{\sqrt{\sum_{i=1}^{n} A_i^2} \sqrt{\sum_{i=1}^{n} B_i^2}}
$$

O resultado varia entre:

| Valor | Interpretação |
|--------|---------------|
| **1** | Vetores na mesma direção (muito semelhantes) |
| **0** | Sem relação aparente |
| **-1** | Direções opostas |

> **Dica 💡**
> 
> Recomendo esse [vídeo](https://www.youtube.com/watch?v=e9U0QAFbfLI) caso você queira entender melhor como funciona a similaridade por cosseno.

Na prática, modelos de embeddings normalmente produzem valores entre **0** e **1**, pois textos semanticamente semelhantes tendem a apontar para direções parecidas no espaço vetorial.

### Similaridade por distância euclidiana

Outra forma bastante utilizada para comparar embeddings é a **Distância Euclidiana**.

Enquanto a Similaridade por Cosseno compara a direção dos vetores, a Distância Euclidiana mede **o espaço físico entre eles**.

Ela corresponde exatamente à distância em linha reta entre dois pontos.

A ●--------------------------● B

Quanto menor essa distância, mais semelhantes tendem a ser os vetores.

Sua fórmula é:

$$d = \sqrt{(x_2 - x_1)^2 + (y_2 - y_1)^2}$$

Em duas dimensões, ela corresponde ao conhecido Teorema de Pitágoras.

Para embeddings, o mesmo conceito é aplicado em centenas ou milhares de dimensões.

Ao contrário da Similaridade por Cosseno, a Distância Euclidiana não possui um intervalo fixo de valores.

Ela varia de:

- **0** → vetores idênticos;
- valores positivos → quanto maior, mais distantes estão os vetores.

Isso significa que seu valor depende da escala dos embeddings produzidos pelo modelo.

### Similaridade por Cosseno × Distância Euclidiana

Embora ambas sejam utilizadas para comparar embeddings, elas medem características diferentes.

| Similaridade por Cosseno | Distância Euclidiana |
|--------------------------|----------------------|
| Compara o ângulo entre vetores | Compara a distância entre vetores |
| Mede a direção | Mede a posição |
| Não é sensível ao comprimento do vetor | É sensível ao comprimento do vetor |
| Intervalo entre -1 e 1 | Intervalo de 0 até infinito |
| Muito utilizada em NLP e IA Generativa | Muito utilizada em visão computacional, clustering e busca vetorial |

Uma forma intuitiva de visualizar essa diferença é imaginar duas pessoas caminhando.

A Similaridade por Cosseno responde:

> **"Elas estão caminhando na mesma direção?"**

Já a Distância Euclidiana responde:

> **"Quão longe uma está da outra?"**

São perguntas diferentes e, portanto, podem produzir respostas diferentes para os mesmos vetores.

### Quando utilizar cada métrica?

Não existe uma métrica universalmente melhor.

A escolha depende da natureza dos embeddings e do problema que estamos resolvendo.

#### Similaridade por Cosseno

É geralmente preferida quando:

- queremos comparar significado semântico visto que sabemos ou não se as palavras (vetores) "estão indo na mesma direção";
- utilizamos embeddings de modelos de linguagem (BERT, bge, e5, OpenAI, Granite etc.);
- o comprimento do vetor não deve influenciar a comparação.

#### Distância Euclidiana

É mais indicada quando:

- a magnitude dos vetores possui significado;
- queremos medir proximidade geométrica;
- utilizamos algoritmos clássicos de clustering ou vizinhos mais próximos.

Em aplicações de IA Generativa e RAG, a Similaridade por Cosseno costuma ser a escolha mais comum, pois modelos de embeddings são treinados para posicionar textos semanticamente semelhantes em direções próximas dentro do espaço vetorial.

````python
from sklearn.metrics.pairwise import cosine_similarity
````

````python
def embed(text: str) -> list[float]:
    return embedding_model.embed_query(text)
````

Agora vamos criar uma base de conhecimento sobre o universo Star Wars e mais umas coisinhas. A ideia é que possamos fazer perguntas sobre esse universo e o modelo de embeddings consiga nos retornar respostas relevantes.

> **Dica 💡**
> 
> O modelo de embeddings já possui um conhecimento prévio ("de fábrica"), lembre-se disso pois dependendo da pergunta que você fizer, o modelo pode te retornar uma resposta mesmo sem você ter adicionado essa informação na base de conhecimento.

````python
knowledge = {
    "Luke Skywalker": embed("Luke Skywalker é um Jedi treinado por Obi-Wan Kenobi e Yoda."),
    "Yoda": embed("Yoda é um Mestre Jedi que treinou Luke Skywalker e outros Jedi."),
    "Obi-Wan Kenobi": embed("Obi-Wan Kenobi é um Mestre Jedi que treinou Luke Skywalker e Anakin Skywalker."),
    "Leia Organa": embed("Leia Organa é uma líder da Aliança Rebelde, irmã de Luke Skywalker."),
    "Darth Vader": embed("Darth Vader é um Lorde Sith e pai de Luke Skywalker. Ele era Anakin Skywalker."),
    "Death Star": embed("Death Star é uma estação espacial destrutiva do Império Galáctico."),
    "Millennium Falcon": embed("Millennium Falcon é uma nave espacial usada pela Aliança Rebelde."),
    "Tatooine": embed("Tatooine é um planeta do universo de Star Wars, foi o lar de Luke Skywalker e Anakin Skywalker."),
    "Jedi": embed("Jedi é um membro da ordem jedi."),
    "Sith": embed("Sith é um membro da ordem sith."),
    "Pizza": embed("Pizza é um prato tradicional italiano feito com massa, molho de tomate e queijo."),
    "Linux": embed("Linux é um sistema operacional de código aberto."),
    "Kubernetes": embed("Kubernetes é uma plataforma de orquestração de contêineres."),
    "Red Hat": embed("Red Hat é uma empresa de software livre.")
}
````

````python
def search(embedded_query: list[float], top_k: int = 5):
    # one line 🗿🤫🧏‍♂️🥶
    # similarities = {key: cosine_similarity([embedded_query], [value])[0][0] for key, value in knowledge.items()}
    
    similarities = {}
    for key, value in knowledge.items():
        similarity_raw = cosine_similarity([embedded_query], [value])
        similarity = similarity_raw[0][0]
        similarities[key] = similarity 

    sorted_results = sorted(similarities.items(), key=lambda item: item[1], reverse=True)
    if top_k is None:
        return sorted_results
    return sorted_results[:top_k]
````

Qual forma de similaridade você acha que estamos usando na função acima? Se você disse similaridade por cosseno está correto.

````python
search(embed("Quem treinou Luke Skywalker?"), top_k=5)
````

Não seria dahora ver isso plotado num gráfico? Bora então !!!

````python
from sklearn.decomposition import PCA
import numpy as np
import pandas as pd
import plotly.express as px

def display_similarities(embedded_query: list[float], query_as_text: str):
    # Embeddings da base
    labels = list(knowledge.keys())
    X = np.array(list(knowledge.values()))

    # Cria e ajusta o PCA
    pca = PCA(n_components=2)
    X_2d = pca.fit_transform(X)

    # Gera o embedding da consulta
    query_embedding = np.array(embedded_query)

    # Projeta a consulta no MESMO espaço do PCA
    query_2d = pca.transform([query_embedding])

    # Monta um DataFrame para o Plotly
    df_knowledge_base = pd.DataFrame({
        "x": X_2d[:, 0],
        "y": X_2d[:, 1],
        "label": labels,
        "type": "Knowledge Base"
    })

    # Adiciona a consulta
    df_query = pd.DataFrame({
        "x": [query_2d[0][0]],
        "y": [query_2d[0][1]],
        "label": [query_as_text],
        "type": ["Query"]
    })

    df = pd.concat([df_knowledge_base, df_query], ignore_index=True)

    fig = px.scatter(
        df,
        x="x",
        y="y",
        text="label",
        color="type",
        title="Embedding Space (PCA)",
        color_discrete_map={
            "Knowledge Base": "royalblue",
            "Query": "red"
        }
    )

    fig.update_traces(textposition="top center")
    fig.update_layout(
        xaxis_title="PCA Component 1",
        yaxis_title="PCA Component 2"
    )

    return fig
````

````python
query_as_embeddings = embed("Quem treinou Luke Skywalker?") 
display_similarities(embedded_query=query_as_embeddings, query_as_text="Quem treinou Luke Skywalker?")
````

E se a gente mudasse a pergunta? O que aconteceria?

````python
query_as_embeddings = embed("OpenShift") 
graph = display_similarities(embedded_query=query_as_embeddings, query_as_text="OpenShift")

similarities = search(embed("OpenShift"), top_k=5)
for label, similarity in similarities:
    print(f"{label}: {similarity:.4f}")

graph
````

````python
query_as_embeddings = embed("Macarrão") 
graph = display_similarities(embedded_query=query_as_embeddings, query_as_text="Macarrão")

similarities = search(embed("Macarrão"), top_k=5)
for label, similarity in similarities:
    print(f"{label}: {similarity:.4f}")

graph
````

Até aqui, vimos que modelos de embeddings transformam textos em vetores numéricos.

Por exemplo, um texto pode ser representado por um vetor com centenas ou milhares de dimensões:

> "Yoda"
> 
>     ↓
> 
> [0.12, -0.34, 0.87, ..., 0.21]
> 
>               ↑
> 
>         1024 dimensões

O problema é que não conseguimos visualizar diretamente um espaço com 1024 dimensões.

Para conseguirmos representar esses embeddings em um gráfico 2D, podemos utilizar uma técnica chamada PCA (Principal Component Analysis), ou Análise de Componentes Principais.

### O que é PCA?

PCA é uma técnica de redução de dimensionalidade.

Seu objetivo é transformar dados que possuem muitas dimensões em uma representação com menos dimensões, tentando preservar a maior quantidade possível da informação presente nos dados.

No nosso caso:

```
Embeddings
     │
     │ 1024 dimensões
     ▼
    PCA
     │
     │ 2 dimensões
     ▼
  Gráfico 2D
```

Assim, podemos transformar:

`14 embeddings × 1024 dimensões`

em:

`14 embeddings × 2 dimensões`

e visualizar esses pontos utilizando um gráfico tradicional de X e Y.

### O que é preservar informações ?

Para entender isso, precisamos entender o conceito de variância.

No contexto do PCA, **variância é uma medida de quanto os dados se espalham em torno da média**.

Considere os dois conjuntos:

```
Conjunto A:

4   5   5   5   6
```

Os valores estão bastante próximos.

Agora:

```
Conjunto B:

1   3   5   7   9
```

Os valores estão muito mais espalhados. O conjunto B possui maior variância. Matematicamente, a variância mede o quanto cada valor se distancia da média:

$$
Var(X) = \frac{1}{N}\sum_{i=1}^{N}(x_i-\mu)^2
$$

onde:

* $(x_i)$ é cada valor;
* $(\mu)$ é a média;
* $(N)$ é a quantidade de valores.

De forma intuitiva:

> Quanto mais espalhados estão os dados, maior a variância.

#### Um resumão

![Resumo Variância e Desvio Padrão](imgs/0a0b1f9b-8103-47c6-8c3c-d3c2d6a1d7ed.png)

### Variância em um espaço de embeddings

Agora imagine que temos vários embeddings representados em um espaço.

             ●
           ●
         ●
       ●
     ●

Existe claramente uma direção em que os dados apresentam bastante variação:

             ●
           ↗
         ●
       ↗
     ●

Essa direção contém uma grande parte da variação existente nos dados.

Então, no fim das contas, o PCA procura justamente por essas direções onde a maior variação está localizada.

### Componentes Principais

O PCA encontra novas direções chamadas de Componentes Principais (Principal Components). O primeiro componente principal, PC1, é a direção que explica a maior quantidade possível da variância dos dados (como vimos na subseção anterior). Depois disso, o PCA encontra uma segunda direção, PC2, perpendicular à primeira, que explica a maior quantidade possível da variância restante.

Visualmente:
```
                    PC2
                     ↑
                     |
                     |
                     |
─────────────────────┼────────────────────→ PC1
                     |
                     |
                     |
```

De forma simplificada:
* PC1 → maior quantidade de variância
* PC2 → segunda maior quantidade de variância
* PC3 → terceira maior quantidade de variância

> ℹ️ **Curiosidade**
>
> Geralmente são utilizados somente os dois primeiros componentes principais (PC1 e PC2) para visualização, mas o PCA pode ser estendido para mais dimensões, caso seja necessário.

### PCA aplicado aos nossos embeddings

Nosso modelo gera embeddings com 1024 dimensões:

```
Embedding
│
├── Dimensão 1
├── Dimensão 2
├── Dimensão 3
├── ...
└── Dimensão 1024
```

Não conseguimos visualizar esse espaço diretamente (seria muito louco pensar em 1024 dimensões 😲).

O PCA cria uma nova representação dos dados:

```
1024 dimensões
      │
      ▼
     PCA
      │
      ├── PC1
      └── PC2
      │
      ▼
   Espaço 2D
```

Agora cada embedding pode ser representado por apenas dois valores.

**Antes:**
```
Yoda
↓
[2.31, -1.42]
```

**Depois:**

O primeiro valor representa a posição do embedding no eixo PC1. O segundo representa sua posição no eixo PC2.

Podemos então visualizar esses pontos:
```

                 PC2
                  ↑
          Yoda ●  |
                  |
      Luke ●      |      ● Obi-Wan
                  |
                  |                    ● Pizza
                  +──────────────────────────→ PC1
```

### Atenção ⚠️ - Observações importantes

Você poderia se perguntar agora: existe algum significado semântico para os eixos PC1 e PC2?

Não !!! Os eixos `PC1` e `PC2` não possuem necessariamente um significado semântico interpretável.

Por exemplo:

> PC1 = "personagens de Star Wars"
> 
> PC2 = "planetas"

não é necessariamente verdade.

O PCA não sabe o que é Star Wars, Yoda, Pizza ou Kubernetes. Ele apenas analisa matematicamente os dados e encontra as direções que explicam a maior quantidade de variância.

Portanto:

> PC1 e PC2 representam direções matemáticas no espaço reduzido, e não conceitos humanos específicos.

Além do significado semântico um questionamento que é comum é confundir o PCA com as métricas utilizadas na busca semântica, nesse caso estou pensando na similaridade por cosseno como exemplo. Pra isso ficar claro, ponha uma coisa na sua cabeça.

O PCA tem como objetivo:

> Reduzir a dimensionalidade para facilitar a análise e visualização dos dados.

Já a similaridade por cosseno responde:

> Quão semelhantes são dois embeddings?
